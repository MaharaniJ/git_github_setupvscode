```python
def revenue_by_status(order_list, status):
    total = 0
    for order in order_list:
        if order["status"] == status:
            total += order["amount"]
    return total
```

```python
import csv
import json
# Our 8 food-delivery orders
orders = [
    {
        "order_id": 101, "customer": "  anita ",
        "restaurant": " spice route", "items": 3,
        "amount": 540.0, "status": "delivered"
    },
    {
        "order_id": 102, "customer": "RAVI",
        "restaurant": "burger hub", "items": 2,
        "amount": 410.0, "status": "delivered"
    }
]
# Create orders.csv
with open("orders.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=orders[0].keys())
    writer.writeheader()
    writer.writerows(orders)
# Create orders.json
with open("orders.json", "w", encoding="utf-8") as f:
    json.dump(orders, f, indent=2)
# Create a plain-text receipt
receipt = """Order #101
Customer: Anita
Restaurant: Spice Route
Items: 3
Amount: Rs 540.0
Status: delivered
"""
with open("receipt_101.txt", "w", encoding="utf-8") as f:
    f.write(receipt)
# Create a deliberately broken JSON file
broken_json = """
[
  {"order_id": 201, "customer": "Divya", "status": "delivered",}
]
"""
with open("orders_broken.json", "w", encoding="utf-8") as f:
    f.write(broken_json)
print("Setup complete!")
print("Created: orders.csv")
print(type(receipt))
print("Created: orders.json")
print("Created: receipt_101.txt")
print("Created: orders_broken.json")
```

```python
import json

def load_orders_safely(filename):
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return json.load(f)

    except FileNotFoundError:
        print(f"Could not find {filename}. Using an empty order list instead.")
        return []

```

```python
result1 = load_orders_safely("orders.json")
print("Loaded:", len(result1), "orders")

result2 = load_orders_safely("january_orders.json")
print("Loaded:", len(result2), "orders")
```

```python
import json
def load_orders_or_warn(filename):
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"Could not find {filename}.")
        return []
    except json.JSONDecodeError as e:
        print(f"{filename} is not valid JSON: {e}")
        return []
```

```python
def load_bad(filename):
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return json.load(f)
    except:
        return "Something went wrong"
```

```python
n = int(input("Enter your number : "))
reversed_num = 0
while n > 0:
  digit = n % 10
  reversed_num = (reversed_num * 10) + digit
  n = n // 10
  print("reversed number", reversed_num)
```

```python
n = int(input("Enter the number of terms: "))
a = 0
b = 1
print("Fibonacci series:")
for i in range(n):
    print(a, end=" ")
    next_term = a + b
    a = b
    b = next_term
```
